## Step 2 — Auto-labelling (Silver Dataset)

We use a **hybrid labeller** that combines:
- `underthesea.ner()` for generic NER (catches PERSON, LOCATION, ORGANIZATION, TIME)
- Domain-specific regex patterns for MONEY, DATE, RATE, PENALTY, LAW
- Keyword dictionaries for PARTY labels

The output of this step is a list of sentences annotated in **BIO token-level** format, ready for fine-tuning.

````python
# ============================================================
# STEP 2 — Load pre-labelled dataset from disk
# ============================================================
import json, time

LABELLED_FILE = '/kaggle/input/datasets/nguynlmhuy/ner-data/labelled_ner_data.json'
# ↑ Change this to wherever you uploaded labelled_ner_data.json
# e.g. Kaggle Dataset path: /kaggle/input/<dataset-name>/labelled_ner_data.json

print(f"Loading labelled dataset from: {LABELLED_FILE}")
t_start = time.time()

with open(LABELLED_FILE, "r", encoding="utf-8") as f:
    labelled = json.load(f)

elapsed = time.time() - t_start
print(f"✓ Loaded {len(labelled):,} labelled examples in {elapsed:.1f}s")

# Validate structure of first few records
print("\nValidating format...")
errors = 0
for i, ex in enumerate(labelled[:1000]):     # spot-check first 1K
    if "tokens" not in ex or "ner_tags" not in ex:
        print(f"  ✗ Missing keys at index {i}")
        errors += 1
    elif len(ex["tokens"]) != len(ex["ner_tags"]):
        print(f"  ✗ Length mismatch at index {i}: "
              f"tokens={len(ex['tokens'])} ner_tags={len(ex['ner_tags'])}")
        errors += 1

if errors == 0:
    print("✓ Format valid (spot-checked first 1,000 examples)")
else:
    print(f"✗ Found {errors} format errors — fix before continuing")

# Entity distribution
from collections import Counter
tag_counts = Counter()
for ex in labelled:
    for tag in ex["ner_tags"]:
        if tag != "O":
            tag_counts[tag.replace("B-","").replace("I-","")] += 1

print("\nEntity distribution (full dataset):")
for label, cnt in sorted(tag_counts.items(), key=lambda x: -x[1]):
    print(f"  {label:<10} {cnt:>10,}")

# Sample
print("\nSample records:")
for ex in labelled[:2]:
    print(f"\n  Tokens  : {ex['tokens'][:8]}")
    print(f"  Tags    : {ex['ner_tags'][:8]}")
````

````python
# ============================================================
# STEP 3 — Filter + balance → 100K train / 60K test
# ============================================================
import random
from collections import Counter, defaultdict
import copy

random.seed(42)

TARGETS = {
    "LAW"    : 100_000,
    "PARTY"  :  36_000,
    "MONEY"  :  36_000,
    "PENALTY":  36_000,
    "DATE"   :  36_000,
    "RATE"   :  33_042,
}

TRAIN_SIZE = 100_000
TEST_SIZE  = 60_000
TOTAL      = TRAIN_SIZE + TEST_SIZE

def get_entity_types(ex):
    types = set()
    for tag in ex["ner_tags"]:
        if tag.startswith("B-"):
            types.add(tag[2:])
    return types


all_labelled = copy.copy(labelled)   
random.shuffle(all_labelled)


raw_split    = int(len(all_labelled) * 0.67)
raw_train    = all_labelled[:raw_split]
raw_test     = all_labelled[raw_split:]

print(f"Raw split before balancing:")
print(f"  raw_train : {len(raw_train):,}")
print(f"  raw_test  : {len(raw_test):,}")


def build_balanced_set(pool, targets, max_size, label):
    """Sample from pool respecting targets, dedup by token string."""
    buckets  = defaultdict(list)
    all_O    = []

    for ex in pool:
        types = get_entity_types(ex)
        if not types:
            all_O.append(ex)
        else:
            for t in types:
                buckets[t].append(ex)

    seen   = set()
    merged = []

    for entity in sorted(targets.keys(), key=lambda k: targets[k]):
        bucket = buckets.get(entity, [])
        random.shuffle(bucket)
        target = targets[entity]
        scaled = int(target * (max_size / TOTAL))
        for ex in bucket[:scaled]:
            key = " ".join(ex["tokens"])
            if key not in seen:
                seen.add(key)
                merged.append(ex)

    # Pad with O-only if needed
    remaining = max_size - len(merged)
    if remaining > 0:
        random.shuffle(all_O)
        for ex in all_O:
            key = " ".join(ex["tokens"])
            if key not in seen:
                seen.add(key)
                merged.append(ex)
            if len(merged) >= max_size:
                break

    random.shuffle(merged)
    print(f"\n  {label} set: {len(merged):,} examples")
    return merged[:max_size]

train_data = build_balanced_set(raw_train, TARGETS, TRAIN_SIZE, "Train")
test_data  = build_balanced_set(raw_test,  TARGETS, TEST_SIZE,  "Test")

# ── Step 3: Verify NO overlap between train and test ──
train_keys = set(" ".join(ex["tokens"]) for ex in train_data)
test_keys  = set(" ".join(ex["tokens"]) for ex in test_data)
overlap    = train_keys & test_keys

print(f"\n✓ Overlap check: {len(overlap)} shared examples between train/test")
if overlap:
    print(f"  ⚠ WARNING: {len(overlap)} overlapping examples found!")
else:
    print(f"  ✓ No data leakage detected")

print(f"\n✓ Final dataset:")
print(f"  Train : {len(train_data):,}")
print(f"  Test  : {len(test_data):,}")

# Entity distribution
for split_name, split_data in [("Train", train_data), ("Test", test_data)]:
    counts = Counter()
    for ex in split_data:
        for tag in ex["ner_tags"]:
            if tag.startswith("B-"):
                counts[tag[2:]] += 1
    print(f"\nEntity distribution ({split_name}):")
    for label in TARGETS:
        print(f"  {label:<10} {counts.get(label, 0):>8,}")
````

## Step 4 — Fine-tune PhoBERT for Token Classification

We use `vinai/phobert-base-v2` with a linear token-classification head.

Key design choices:
- **Syllable-level tokenisation** — PhoBERT expects VNCoreNLP-style tokens; we supply word-tokenized text (underthesea) which is close enough.
- **Label alignment** — For sub-word tokens produced by the BPE tokenizer, we propagate the first sub-word label and set remaining sub-words to `-100` (ignored in loss).
- **Batch size** — Use gradient accumulation to simulate large batch on limited VRAM.

````python
# ============================================================
# STEP 4a — Define label scheme
# ============================================================

ENTITY_TYPES = ["PARTY", "MONEY", "DATE", "RATE", "PENALTY", "LAW"]

LABEL_LIST = ["O"]
for ent in ENTITY_TYPES:
    LABEL_LIST.append(f"B-{ent}")
    LABEL_LIST.append(f"I-{ent}")

LABEL2ID = {lbl: i for i, lbl in enumerate(LABEL_LIST)}
ID2LABEL = {i: lbl for lbl, i in LABEL2ID.items()}

print(f"✓ Label scheme ({len(LABEL_LIST)} labels):")
for lbl, idx in LABEL2ID.items():
    print(f"  {idx:>2}  {lbl}")
````

````python
# ============================================================
# STEP 4b — Build HuggingFace Dataset
# ============================================================
from datasets import Dataset, DatasetDict

def to_hf_format(examples):
    """Convert our dict list to HF Dataset format with integer labels."""
    out = {"tokens": [], "ner_tags": []}
    for ex in examples:
        out["tokens"].append(ex["tokens"])
        out["ner_tags"].append([LABEL2ID.get(t, 0) for t in ex["ner_tags"]])
    return out

train_dict = to_hf_format(train_data)
test_dict  = to_hf_format(test_data)

ds = DatasetDict({
    "train": Dataset.from_dict(train_dict),
    "test" : Dataset.from_dict(test_dict),
})

print("✓ HuggingFace DatasetDict:")
print(ds)
print("\nSample train example:")
ex = ds["train"][0]
for tok, tag_id in zip(ex["tokens"][:10], ex["ner_tags"][:10]):
    print(f"  {tok:<20}  {ID2LABEL[tag_id]}")
````

````python
from transformers import AutoTokenizer

MODEL_NAME = "vinai/phobert-base-v2"
tokenizer  = AutoTokenizer.from_pretrained(MODEL_NAME, use_fast=False)

print(f"✓ Tokenizer loaded : {MODEL_NAME}")
print(f"  Fast tokenizer   : {tokenizer.is_fast}")
print(f"  Vocab size       : {tokenizer.vocab_size:,}")

sample = "Bên B phải thanh toán 10.000.000 VNĐ."
enc = tokenizer(sample.split(), is_split_into_words=True)
print(f"\nEncoding test on: '{sample}'")
print(f"  Subword tokens: {tokenizer.convert_ids_to_tokens(enc['input_ids'])}")
````

````python
# ============================================================
# STEP 4d — Tokenize & align labels (slow-tokenizer safe)
# ============================================================
# Strategy: tokenize each word individually to know exactly how many
# sub-word pieces each word produces, then assign:
#   • first piece  → original label
#   • further pieces + special tokens → -100 (ignored in loss)
# ============================================================
MAX_SEQ_LEN = 256

def tokenize_and_align_labels(examples):
    """
    Slow-tokenizer-compatible label alignment for PhoBERT.

    For each sentence we:
      1. Tokenize word-by-word (no special tokens) to count sub-word pieces.
      2. Tokenize the full word list (with special tokens, truncated).
      3. Walk the sub-word count list to assign labels, skipping [CLS]/[SEP].
    """
    all_input_ids      = []
    all_attention_mask = []
    all_labels         = []

    for words, orig_labels in zip(examples["tokens"], examples["ner_tags"]):

        # --- Step 1: count sub-word pieces per word ---
        pieces_per_word = [
            len(tokenizer.tokenize(w)) or 1   # guard against empty
            for w in words
        ]

        # --- Step 2: full encoding with truncation ---
        enc = tokenizer(
            words,
            is_split_into_words=True,
            truncation=True,
            max_length=MAX_SEQ_LEN,
            padding=False,
        )
        input_ids      = enc["input_ids"]
        attention_mask = enc["attention_mask"]

        # --- Step 3: build aligned label list ---
        # [CLS] → -100
        labels = [-100]
        token_budget = MAX_SEQ_LEN - 2   # reserve [CLS] + [SEP]
        used = 0

        for word_idx, n_pieces in enumerate(pieces_per_word):
            if used + n_pieces > token_budget:
                break                          # truncated
            word_label = orig_labels[word_idx] if word_idx < len(orig_labels) else 0
            labels.append(word_label)          # first sub-word → real label
            labels.extend([-100] * (n_pieces - 1))  # rest → ignored
            used += n_pieces

        # [SEP] → -100
        labels.append(-100)

        # Sanity: lengths must match
        assert len(labels) == len(input_ids), (
            f"Length mismatch: labels={len(labels)}, tokens={len(input_ids)}"
        )

        all_input_ids.append(input_ids)
        all_attention_mask.append(attention_mask)
        all_labels.append(labels)

    return {
        "input_ids"      : all_input_ids,
        "attention_mask" : all_attention_mask,
        "labels"         : all_labels,
    }


print("Tokenizing train set (slow-tokenizer mode, this may take a few minutes)...")
tokenized_ds = ds.map(
    tokenize_and_align_labels,
    batched=True,
    batch_size=256,              # smaller batch → lower peak RAM
    remove_columns=ds["train"].column_names,
    desc="Tokenizing",
)

print("\n✓ Tokenized dataset:")
print(tokenized_ds)

# Spot-check alignment on first example
ex   = tokenized_ds["train"][0]
toks = tokenizer.convert_ids_to_tokens(ex["input_ids"])
print("\nSpot-check (first 15 sub-word positions):")
print(f"  {'Token':<20} Label")
for tok, lbl in list(zip(toks, ex["labels"]))[:15]:
    tag = ID2LABEL.get(lbl, "[IGN]") if lbl != -100 else "[IGN]"
    print(f"  {tok:<20} {tag}")
````

````python
# ============================================================
# STEP 4e — Load PhoBERT model with NER head
# ============================================================
from transformers import AutoModelForTokenClassification

model = AutoModelForTokenClassification.from_pretrained(
    MODEL_NAME,
    num_labels=len(LABEL_LIST),
    id2label=ID2LABEL,
    label2id=LABEL2ID,
)

total_params = sum(p.numel() for p in model.parameters())
trainable    = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"✓ Model loaded: {MODEL_NAME}")
print(f"  Total params    : {total_params/1e6:.1f} M")
print(f"  Trainable params: {trainable/1e6:.1f} M")
````

````python
# ============================================================
# STEP 4f — Metrics (seqeval)
# ============================================================
import numpy as np
from seqeval.metrics import classification_report, f1_score

def compute_metrics(eval_preds):
    logits, labels = eval_preds
    predictions = np.argmax(logits, axis=-1)

    true_labels  = []
    true_preds   = []

    for pred_seq, label_seq in zip(predictions, labels):
        true_labels.append([
            ID2LABEL[l] for p, l in zip(pred_seq, label_seq) if l != -100
        ])
        true_preds.append([
            ID2LABEL[p] for p, l in zip(pred_seq, label_seq) if l != -100
        ])

    report = classification_report(true_labels, true_preds, output_dict=True)

    return {
        "f1"       : report.get("micro avg", {}).get("f1-score", 0.0),
        "precision": report.get("micro avg", {}).get("precision", 0.0),
        "recall"   : report.get("micro avg", {}).get("recall", 0.0),
    }

print("✓ Metrics function defined (seqeval micro-F1)")
````

````python
from transformers import TrainingArguments, Trainer, DataCollatorForTokenClassification

MODEL_SAVE_DIR = "/kaggle/working/phobert-ner-legal"

training_args = TrainingArguments(
    output_dir                  = MODEL_SAVE_DIR,
    num_train_epochs            = 10,
    per_device_train_batch_size = 16,
    per_device_eval_batch_size  = 32,
    gradient_accumulation_steps = 1,
    learning_rate               = 1e-5,
    weight_decay                = 1e-2,
    warmup_steps                = 300,
    lr_scheduler_type           = "linear",

    eval_strategy               = "epoch",      
    save_strategy               = "epoch",     
    save_total_limit            = 1,           
    load_best_model_at_end      = True,
    metric_for_best_model       = "f1",

    logging_steps               = 500,
    report_to                   = "none",
    fp16                        = True,
    ddp_find_unused_parameters  = False,
    dataloader_num_workers      = 2,
    seed                        = 42,
)

data_collator = DataCollatorForTokenClassification(tokenizer)

trainer = Trainer(
    model            = model,
    args            = training_args,
    train_dataset   = tokenized_ds["train"],
    eval_dataset    = tokenized_ds["test"],
    processing_class= tokenizer,
    data_collator   = data_collator,
    compute_metrics = compute_metrics,
)

print("✓ Trainer ready")
print(f"  Effective batch size : {training_args.per_device_train_batch_size * training_args.gradient_accumulation_steps}")
print(f"  Epochs               : {training_args.num_train_epochs}")
print(f"  LR                   : {training_args.learning_rate}")
````

````python
# ============================================================
# STEP 4h — TRAIN
# ============================================================
print("Starting training...")
train_result = trainer.train()

print("\n" + "="*55)
print("TRAINING COMPLETE")
print("="*55)
print(f"  Runtime         : {train_result.metrics['train_runtime']:.0f} s")
print(f"  Samples/sec     : {train_result.metrics['train_samples_per_second']:.1f}")
print(f"  Final train loss: {train_result.metrics['train_loss']:.4f}")

# Save model
trainer.save_model(MODEL_SAVE_DIR)
tokenizer.save_pretrained(MODEL_SAVE_DIR)
print(f"\n✓ Model saved to: {MODEL_SAVE_DIR}")
````

````python
# ============================================================
# STEP 5 — Evaluate on test set
# ============================================================
import numpy as np
from seqeval.metrics import classification_report as seq_report

print("Running evaluation on test set...")
eval_results = trainer.evaluate(tokenized_ds["test"])

print("\n" + "="*55)
print("TEST SET RESULTS")
print("="*55)
print(f"  Micro F1    : {eval_results['eval_f1']:.4f}")
print(f"  Precision   : {eval_results['eval_precision']:.4f}")
print(f"  Recall      : {eval_results['eval_recall']:.4f}")
print(f"  Eval loss   : {eval_results['eval_loss']:.4f}")

# Per-entity breakdown
print("\nPer-entity breakdown:")
predictions_output = trainer.predict(tokenized_ds["test"])
preds  = np.argmax(predictions_output.predictions, axis=-1)
labels = predictions_output.label_ids

true_labels = []
true_preds  = []
for pred_seq, label_seq in zip(preds, labels):
    true_labels.append([ID2LABEL[l] for p, l in zip(pred_seq, label_seq) if l != -100])
    true_preds.append( [ID2LABEL[p] for p, l in zip(pred_seq, label_seq) if l != -100])

print(seq_report(true_labels, true_preds))
````

````python
import torch
import torch.distributed as dist

if dist.is_available() and dist.is_initialized():
    dist.destroy_process_group()
    print("✓ DDP process group destroyed")

# Force single GPU for inference
os.environ["CUDA_VISIBLE_DEVICES"] = "0"
os.environ["CUDA_LAUNCH_BLOCKING"] = "1"
torch.cuda.empty_cache()
print("✓ Inference will run on GPU 0 only")
print(f"  GPU memory free: {torch.cuda.mem_get_info()[0]/1e9:.1f} GB")
````

````python
# ============================================================
# STEP 6a — NER inference pipeline (single GPU, safe mode)
# ============================================================
import os, torch
import torch.distributed as dist
from transformers import AutoModelForTokenClassification, AutoTokenizer, pipeline

os.environ["CUDA_LAUNCH_BLOCKING"] = "1"
os.environ["CUDA_VISIBLE_DEVICES"] = "0"

# Destroy DDP if still active
if dist.is_available() and dist.is_initialized():
    dist.destroy_process_group()
    print("✓ DDP destroyed")

torch.cuda.empty_cache()

# Reload model — explicitly FP32 to avoid precision mismatch
infer_model = AutoModelForTokenClassification.from_pretrained(
    MODEL_SAVE_DIR,
    torch_dtype = torch.float32,    # ← explicit FP32, no mixed precision issues
)
infer_model.eval()                  # ← set eval mode explicitly

infer_tokenizer = AutoTokenizer.from_pretrained(
    MODEL_SAVE_DIR,
    use_fast = False
)

# ── Do NOT set batch_size here — we manually batch in Cell 20 ──
ner_pipeline = pipeline(
    task                 = "ner",
    model                = infer_model,
    tokenizer            = infer_tokenizer,
    aggregation_strategy = "simple",
    device               = 0,
)

print("✓ NER pipeline ready")
print(f"  Model dtype : {next(infer_model.parameters()).dtype}")
print(f"  Device      : {next(infer_model.parameters()).device}")

# Sanity check on 3 sentences
demo = [
    "Bên B phải thanh toán 10.000.000 VNĐ trước ngày 05/5/2024.",
    "Nếu thanh toán trễ hạn, mức phạt 1% mỗi ngày sẽ được áp dụng.",
    "Hợp đồng được ký kết theo quy định của Bộ luật Dân sự 2015.",
]
print("\nSanity check:")
for sent in demo:
    ents = ner_pipeline(sent)        # single string, no batching
    print(f"\n  Clause: {sent}")
    for e in ents:
        print(f"    [{e['entity_group']}] '{e['word']}'  score={e['score']:.3f}")
````

````python
# ============================================================
# STEP 6b — Run inference → ner_results.json
# ============================================================
import json, os, time

# ── Reload variables lost after training ──
CLAUSE_FILE  = "/kaggle/working/output/clauses.txt"
OUTPUT_DIR   = "/kaggle/working/output"
NER_OUTPUT   = f"{OUTPUT_DIR}/ner_results.json"
REPORT_EVERY = 10_000
BATCH_SIZE   = 32

ENTITY_TYPES = ["PARTY", "MONEY", "DATE", "RATE", "PENALTY", "LAW"]
VALID_LABELS = set(ENTITY_TYPES)    # entity_group returns "PARTY" not "B-PARTY"

with open(CLAUSE_FILE, "r", encoding="utf-8") as f:
    all_clauses = [line.strip() for line in f if line.strip()]

print(f"✓ Loaded {len(all_clauses):,} clauses")
if not all_clauses:
    raise RuntimeError(f"clauses.txt is empty at {CLAUSE_FILE}")

# ── Manual batching with plain Python lists ──
# Never pass HFDataset column or Arrow array to pipeline
t_start = time.time()

with open(NER_OUTPUT, "w", encoding="utf-8") as f_out:
    f_out.write("[\n")
    first_entry     = True
    total_processed = 0

    for batch_start in range(0, len(all_clauses), BATCH_SIZE):
        batch = all_clauses[batch_start : batch_start + BATCH_SIZE]

        # Filter out empty strings before passing to pipeline
        batch = [c for c in batch if c.strip()]
        if not batch:
            continue

        try:
            results = ner_pipeline(batch)
        except Exception as ex:
            print(f"  ⚠ Skipping batch {batch_start}–{batch_start+BATCH_SIZE}: {ex}")
            total_processed += len(batch)
            continue

        for clause, ents in zip(batch, results):
            entities = [
                {"text": e["word"].strip(), "label": e["entity_group"]}
                for e in ents
                if e["entity_group"] in VALID_LABELS
                and e["score"] >= 0.70
                and e["word"].strip()
            ]

            if not first_entry:
                f_out.write(",\n")
            f_out.write(json.dumps(
                {"clause": clause, "entities": entities},
                ensure_ascii=False
            ))
            first_entry      = False
            total_processed += 1

        if total_processed % REPORT_EVERY < BATCH_SIZE:
            elapsed = time.time() - t_start
            speed   = total_processed / elapsed
            eta     = (len(all_clauses) - total_processed) / speed
            print(f"  [{total_processed:>9,} / {len(all_clauses):,}]"
                  f"  speed={speed:.0f}/s  ETA={eta/60:.1f}min")

    f_out.write("\n]")

elapsed = time.time() - t_start
print(f"\n✓ Saved: {NER_OUTPUT}")
print(f"  Total processed : {total_processed:,}")
print(f"  Total time      : {elapsed:.0f}s = {elapsed/60:.1f}min")
````

````python
# ============================================================
# STEP 6c — Verify output & show sample
# ============================================================
import json
from collections import Counter

NER_OUTPUT = f"{OUTPUT_DIR}/ner_results.json"

with open(NER_OUTPUT, "r", encoding="utf-8") as f:
    results = json.load(f)

print(f"✓ Total records: {len(results):,}")

label_freq = Counter()
for r in results:
    for e in r["entities"]:
        label_freq[e["label"]] += 1

print("\nEntity label distribution:")
for lbl, cnt in label_freq.most_common():
    print(f"  {lbl:<10}  {cnt:>10,}")

clauses_with_ents = sum(1 for r in results if r["entities"])
print(f"\nClauses with ≥1 entity : {clauses_with_ents:,} "
      f"({100*clauses_with_ents/len(results):.1f}%)")

print("\n" + "="*60)
print("SAMPLE OUTPUT (first 5 clauses with entities)")
print("="*60)
shown = 0
for r in results:
    if r["entities"]:
        print(f"\nClause : {r['clause']}")
        for e in r["entities"]:
            print(f"  [{e['label']}]  {e['text']}")
        shown += 1
        if shown >= 5:
            break

print("\n" + "="*60)
print("JSON FORMAT PREVIEW (first entry with entities)")
print("="*60)
for r in results:
    if r["entities"]:
        print(json.dumps(r, ensure_ascii=False, indent=2))
        break
````
